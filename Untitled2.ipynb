```python
from sklearn.model_selection import train_test_split
import shutil
import random

# Use a fixed seed so we can reproduce the split
random.seed(42)

def split_and_copy(images, label):
    # 70% train, 30% temporary
    train, temp = train_test_split(
        images,
        test_size=0.30,
        random_state=42
    )

    # Split remaining 30% equally into validation and test
    val, test = train_test_split(
        temp,
        test_size=0.50,
        random_state=42
    )

    for image in train:
        shutil.copy2(image, DEST / "train" / label / image.name)

    for image in val:
        shutil.copy2(image, DEST / "val" / label / image.name)

    for image in test:
        shutil.copy2(image, DEST / "test" / label / image.name)

    print(
        label,
        "→",
        "train:", len(train),
        "val:", len(val),
        "test:", len(test)
    )

split_and_copy(healthy_images, "healthy")
split_and_copy(unhealthy_images, "unhealthy")
```

```python
from pathlib import Path
from sklearn.model_selection import train_test_split

BASE = Path("/content/drive/MyDrive/mourchi maushi/New Onion - Copy/2. Bulb")

image_extensions = {".jpg", ".jpeg", ".png", ".webp"}

healthy = []
unhealthy = []

for path in BASE.rglob("*"):
    if path.is_file() and path.suffix.lower() in image_extensions:
        if "1. Healthy" in path.parts:
            healthy.append(path)
        elif "2. Unhealthy" in path.parts:
            unhealthy.append(path)

print("Healthy:", len(healthy))
print("Unhealthy:", len(unhealthy))

# 70% train, 15% validation, 15% test
healthy_train, healthy_temp = train_test_split(
    healthy, test_size=0.30, random_state=42
)

healthy_val, healthy_test = train_test_split(
    healthy_temp, test_size=0.50, random_state=42
)

unhealthy_train, unhealthy_temp = train_test_split(
    unhealthy, test_size=0.30, random_state=42
)

unhealthy_val, unhealthy_test = train_test_split(
    unhealthy_temp, test_size=0.50, random_state=42
)

train_data = [(p, 0) for p in healthy_train] + [(p, 1) for p in unhealthy_train]
val_data   = [(p, 0) for p in healthy_val]   + [(p, 1) for p in unhealthy_val]
test_data  = [(p, 0) for p in healthy_test]  + [(p, 1) for p in unhealthy_test]

print("\nDataset split:")
print("Train:", len(train_data))
print("Validation:", len(val_data))
print("Test:", len(test_data))

print("\nClasses:")
print("0 = Healthy")
print("1 = Unhealthy")
```

```python
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import torch
from torchvision import transforms

class OnionDataset(Dataset):
    def __init__(self, data, transform=None):
        self.data = data
        self.transform = transform

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        path, label = self.data[idx]

        image = Image.open(path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, label
```

```python
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
train_dataset = OnionDataset(train_data, train_transform)
val_dataset = OnionDataset(val_data, eval_transform)
test_dataset = OnionDataset(test_data, eval_transform)

print("Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))
```

```python
image, label = train_dataset[0]

print("Tensor shape:", image.shape)
print("Label:", "Healthy" if label == 0 else "Unhealthy")
```

```python
from torchvision.models import mobilenet_v3_small, MobileNet_V3_Small_Weights
import torch.nn as nn

weights = MobileNet_V3_Small_Weights.DEFAULT

model = mobilenet_v3_small(weights=weights)

# Replace the final classifier for our 2 classes
model.classifier[3] = nn.Linear(
    model.classifier[3].in_features,
    2
)

print(model.classifier)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = model.to(device)

print("Using:", device)
print("GPU:", torch.cuda.get_device_name(0))
```

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True,
    num_workers=4,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=64,
    shuffle=False,
    num_workers=4,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False,
    num_workers=4,
    pin_memory=True
)

print("Batches:")
print("Train:", len(train_loader))
print("Validation:", len(val_loader))
print("Test:", len(test_loader))
```

```python
import torch.optim as optim

criterion = nn.CrossEntropyLoss()

optimizer = optim.AdamW(
    model.parameters(),
    lr=0.0001,
    weight_decay=0.0001
)

print("Optimizer ready.")
```

```python
import torch
import torch.nn as nn
from pathlib import Path

# --------------------------------------------------
# 1. Class weights to compensate for imbalance
# --------------------------------------------------

num_healthy = len(healthy_images)
num_unhealthy = len(unhealthy_images)

total = num_healthy + num_unhealthy

class_weights = torch.tensor([
    total / (2 * num_healthy),      # Healthy
    total / (2 * num_unhealthy)     # Unhealthy
], dtype=torch.float32).to(device)

criterion = nn.CrossEntropyLoss(weight=class_weights)

print("Class weights:", class_weights)


# --------------------------------------------------
# 2. Optimizer
# --------------------------------------------------

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)


# --------------------------------------------------
# 3. Save checkpoints to Google Drive
# --------------------------------------------------

CHECKPOINT_DIR = Path(
    "/content/drive/MyDrive/mourchi maushi/model_checkpoints"
)

CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

best_val_accuracy = 0.0


# --------------------------------------------------
# 4. Training loop
# --------------------------------------------------

EPOCHS = 10

for epoch in range(EPOCHS):

    # -------------------------
    # TRAIN
    # -------------------------
    model.train()

    running_loss = 0.0
    correct = 0
    total_samples = 0

    for images, labels in train_loader:

        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)

        predictions = outputs.argmax(dim=1)

        correct += (predictions == labels).sum().item()
        total_samples += labels.size(0)

    train_loss = running_loss / total_samples
    train_accuracy = correct / total_samples


    # -------------------------
    # VALIDATION
    # -------------------------
    model.eval()

    val_loss_total = 0.0
    val_correct = 0
    val_samples = 0

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_loss_total += loss.item() * images.size(0)

            predictions = outputs.argmax(dim=1)

            val_correct += (predictions == labels).sum().item()
            val_samples += labels.size(0)

    val_loss = val_loss_total / val_samples
    val_accuracy = val_correct / val_samples


    # -------------------------
    # PRINT RESULTS
    # -------------------------

    print(
        f"Epoch [{epoch+1}/{EPOCHS}] "
        f"| Train Loss: {train_loss:.4f} "
        f"| Train Acc: {train_accuracy:.4f} "
        f"| Val Loss: {val_loss:.4f} "
        f"| Val Acc: {val_accuracy:.4f}"
    )


    # -------------------------
    # SAVE BEST MODEL
    # -------------------------

    if val_accuracy > best_val_accuracy:

        best_val_accuracy = val_accuracy

        checkpoint_path = CHECKPOINT_DIR / "best_onion_model.pth"

        torch.save({
            "epoch": epoch + 1,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "val_accuracy": val_accuracy
        }, checkpoint_path)

        print(f"  ✓ New best model saved → {checkpoint_path}")
```

```python
from pathlib import Path
from sklearn.model_selection import train_test_split
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image

BASE = Path("/content/drive/MyDrive/mourchi maushi/New Onion - Copy/2. Bulb")

image_extensions = {".jpg", ".jpeg", ".png", ".webp"}

healthy = []
unhealthy = []

for path in BASE.rglob("*"):
    if path.is_file() and path.suffix.lower() in image_extensions:
        if "1. Healthy" in path.parts:
            healthy.append(path)
        elif "2. Unhealthy" in path.parts:
            unhealthy.append(path)

# Recreate EXACTLY the same split
_, healthy_temp = train_test_split(
    healthy, test_size=0.30, random_state=42
)

healthy_val, healthy_test = train_test_split(
    healthy_temp, test_size=0.50, random_state=42
)

_, unhealthy_temp = train_test_split(
    unhealthy, test_size=0.30, random_state=42
)

unhealthy_val, unhealthy_test = train_test_split(
    unhealthy_temp, test_size=0.50, random_state=42
)

test_data = (
    [(p, 0) for p in healthy_test] +
    [(p, 1) for p in unhealthy_test]
)

# Test images must NOT be augmented
test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

class OnionDataset(Dataset):
    def __init__(self, data, transform=None):
        self.data = data
        self.transform = transform

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        path, label = self.data[idx]

        image = Image.open(path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, label

test_dataset = OnionDataset(test_data, test_transform)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False,
    num_workers=4,
    pin_memory=True
)

print("Test images:", len(test_dataset))
print("Test batches:", len(test_loader))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

model.eval()

all_predictions = []
all_labels = []

with torch.no_grad():
    for images, labels in test_loader:

        images = images.to(device, non_blocking=True)

        outputs = model(images)
        predictions = outputs.argmax(dim=1)

        all_predictions.extend(predictions.cpu().numpy())
        all_labels.extend(labels.numpy())

print("Confusion Matrix:")
print(confusion_matrix(all_labels, all_predictions))

print("\nClassification Report:")
print(
    classification_report(
        all_labels,
        all_predictions,
        target_names=["Healthy", "Unhealthy"],
        digits=4
    )
)
```

```python
import matplotlib.pyplot as plt
from PIL import Image

# Find the mistakes
mistakes = []

model.eval()

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)

        outputs = model(images)
        predictions = outputs.argmax(dim=1)

        for i in range(len(labels)):
            if predictions[i].item() != labels[i].item():
                mistakes.append({
                    "image": images[i].cpu(),
                    "actual": labels[i].item(),
                    "predicted": predictions[i].item()
                })

print("Number of mistakes:", len(mistakes))

# Display them
fig, axes = plt.subplots(1, len(mistakes), figsize=(16, 4))

for ax, mistake in zip(axes, mistakes):
    image = mistake["image"]

    # Undo ImageNet normalization
    image = image.permute(1, 2, 0)
    image = image * torch.tensor([0.229, 0.224, 0.225]) + \
            torch.tensor([0.485, 0.456, 0.406])

    image = image.clamp(0, 1)

    actual = "Healthy" if mistake["actual"] == 0 else "Unhealthy"
    predicted = "Healthy" if mistake["predicted"] == 0 else "Unhealthy"

    ax.imshow(image)
    ax.set_title(f"Actual: {actual}\nPredicted: {predicted}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```
